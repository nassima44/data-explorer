## First look at the data
- The dataset has 56,269 rows (one row per entity and year) and 79 columns.
- `country` and `year` are complete. Most other columns contain missing values (`NaN`), especially for early years.
- 36 entities have no `iso_code`: 35 are not countries (regional aggregates such as `World`, income groups, international aviation and shipping, Kuwaiti oil fires), and one is a real country: Kosovo.

```python
countries = df[(df["iso_code"].notna()) | (df["country"]=="Kosovo")]
```

```python
countries["country"].nunique()
```

```python
countries_1950_2024 = countries[(countries["year"] >= 1950) & (countries["year"] <= 2024)]
```

## Missing values (countries, 1950–2024)
- After filtering, there are 16,800 rows: 224 countries × 75 years. Every country has a row for every year, but many values are missing.
- `co2` and `population` are almost complete (about 7% and 1% missing).
- `gas_co2` (52%), `coal_co2` (37%) and `gdp` (32%) have many missing values. `gdp` stops in 2022.
- `co2_per_capita` has 83 more missing values than `co2`: these rows have `co2` but no `population`, so the per-capita value cannot be computed.
- Missing values mean "unknown", not zero. Whether `NaN` in `gas_co2` sometimes means a real zero still needs to be checked in the codebook.
